```python
# exec("dat_{} = np.genfromtxt(cat_name[0], dtype=None, encoding='ascii', names=colnames)".format(band[0]))
# exec("dat_{} = np.genfromtxt(cat_nam"e[1], dtype=None, encoding='ascii', names=colnames)".format(band[1]))
# Code that makes data name containts wave band
colnames = ['x','y','num','mag_auto','magerr_auto','mag_iso','magerr_iso','mag_isocor','magerr_isocor','kron','backgr','ra','dec', 
            'a','b','theta','mu0','flag','fwhm','flxrad','cl']
dat_f115w_0 = np.genfromtxt("./F115WF150W/F115W.cat", dtype=None, encoding='ascii', names=colnames)
dat_f115w_1 = np.genfromtxt("./F115WF200W/F115W.cat", dtype=None, encoding='ascii', names=colnames)
dat_f150w_0 = np.genfromtxt("./F115WF150W/F150W.cat", dtype=None, encoding='ascii', names=colnames)
dat_f150w_1 = np.genfromtxt("./F150WF200W/F150W.cat", dtype=None, encoding='ascii', names=colnames)
dat_f200w_0 = np.genfromtxt("./F115WF200W/F200W.cat", dtype=None, encoding='ascii', names=colnames)
dat_f200w_1 = np.genfromtxt("./F150WF200W/F200W.cat", dtype=None, encoding='ascii', names=colnames)
```

```python
temp11 = pd.DataFrame(dat_f115w_0)

temp11 = temp11[temp11['cl']>0.8]
temp11 = temp11[temp11['mag_iso']<24]
temp11 = temp11[temp11['fwhm']<0.08*(1/0.03)]

from astropy.io import fits 
dat = fits.getdata("../../img/align-image/jw01345-o002_t022_nircam_clear_f115w_aligned.fits", ext=1)

sz=30
key = [2, 5, 6]

from astropy.visualization import ZScaleInterval

fig, axes = plt.subplots(1, 3, figsize=(10, 4))
fig.suptitle('Sample Point Source of F115W img', fontsize=21)
for i in range(3):
    samp = temp11.iloc[key[i]]
    ZInterval = ZScaleInterval()
    vmin, vmax = ZInterval.get_limits(dat)
    axes[i].imshow(dat, vmin=vmin, vmax=vmax)
    axes[i].set_xlim(samp['x']-sz,samp['x']+sz)
    axes[i].set_ylim(samp['y']-sz,samp['y']+sz)
    axes[i].set_title(f"FWHM : {samp['fwhm']*0.03:.3f} arcsec", fontsize=20)
    axes[i].axis('off')
    
plt.tight_layout()
```
